```python
toolkit = GitHubToolkit.from_github_api_wrapper(
    github, include_release_tools=True
)
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent

# Disponibiliza todas as ferramentas com nomes válidos para o modelo.
import re

tools = toolkit.get_tools()
for tool in tools:
    tool.name = re.sub(r"[^a-zA-Z0-9_-]+", "_", tool.name).strip("_").lower()

llm = ChatOpenAI(model="gpt-4o-mini")
agent_executor = create_agent(model=llm, tools=tools)

example_query = "o que fala o arquivo readme do repositório https://github.com/Leonardojdss/AgentSight-tcc-poli-usp"

events = agent_executor.stream(
    {"messages": [("user", example_query)]},
    stream_mode="values",
)
for event in events:
    event["messages"][-1].pretty_print()
```
